a) (0.6 pts) Crea una función imc(weight, height) que devuelva el índice de masa corporal (IMC), redondeado a dos decimales.



```python
def imc(weight, height):
    """
    Calcula el índice de masa corporal (IMC) dado el peso y la altura.
    
    Parámetros:
    weight (float): Peso en kilogramos.
    height (float): Altura en metros.
    
    Retorna:
    float: IMC redondeado a dos decimales.
    """
    if height <= 0:
        raise ValueError("La altura debe ser mayor que cero.")
    
    imc_value = weight / (height ** 2)
    return round(imc_value, 2)
```

```python
def saludable(imcValue):
    """
    Determina la categoría de salud basada en el IMC.
    
    Parámetros:
    imcValue (float): Valor del índice de masa corporal (IMC).
    
    Retorna:
    str: Categoría de salud ("Saludable", "Bajo peso" o "Sobrepeso").
    """
    if imcValue < 18.5:
        return "Bajo peso"
    elif 18.5 <= imcValue <= 24.9:
        return "Saludable"
    else:
        return "Sobrepeso"
```

d) (0.2 pts) ¿Cómo podrías integrar ambas funciones dentro de una clase llamada HealthTools que calcule el IMC y clasifique automáticamente a partir de listas de pesos y alturas?

```python
class HealthTools:
    
    def __init__(self, weights, heights):
        """
        Inicializa la clase con listas de pesos y alturas.
        
        Parámetros:
        weights (list): Lista de pesos en kilogramos.
        heights (list): Lista de alturas en metros.
        """
        self.weights = weights
        self.heights = heights
        self.imc_values = []
        self.categories = []
    
    def calculate_imc(self):
        for weight, height in zip(self.weights, self.heights):
            imc_value = imc(weight, height)
            self.imc_values.append(imc_value)
    
    def classify_health(self):
        """
        Clasifica la salud basada en los valores de IMC calculados y almacena los resultados en categories.
        """
        for imc_value in self.imc_values:
            category = saludable(imc_value)
            self.categories.append(category)
    
    def get_results(self):
        """
        Devuelve un DataFrame con los pesos, alturas, IMC y categorías de salud.
        
        Retorna:
        pd.DataFrame: DataFrame con las columnas 'peso', 'altura', 'imc' y 'categoria'.
        """
        return pd.DataFrame({
            'peso': self.weights,
            'altura': self.heights,
            'imc': self.imc_values,
            'categoria': self.categories
        })
```
